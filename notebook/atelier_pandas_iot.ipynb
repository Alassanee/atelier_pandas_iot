#### `1.5 Création, de deux façon différentes, d'une Series de 4 températures avec comme index les heures demesure 12h, 13h, 14h et 15h`

```python
# Methode 1: a partir d'un dictionnaire
dict_temp = {
     "12h": 34,
     "13h": 23,
     "14h": 40,
     "15h": 37
}
serie_1 = pd.Series(dict_temp)
serie_1
```

## `Partie 5 – Manipulation des colonnes`

### `5.1 Ajouter la colonne "temperature_fahrenheit"`

```python
df["temperature_fahrenheit"] = (df["temperature"] * 1.8) +32
df["temperature_fahrenheit"]
```

### `5.2 Ajouter la colonne "niveau_temperature" pour catégoriser la température en "Élevée" (température > 30) ou "Normale". `

```python
import numpy as np

df["nouveau_temperature"] = np.where(df["temperature"] > 30, "Elevée", "Normal")
df["nouveau_temperature"]
```

### `5.3 Renommage de la colonne "humidite" en "humidite_relative"`

```python
df = df.rename(columns = {"humidite":"humidite_relative"})
df
```

### `5.4 Suppression de la colonne "temperature_fahrenheit"`

```python
df = df.drop(columns=["temperature_fahrenheit"])
df
```

## `Partie 6 – Filtrage`

#### `6.1 Récupération des lignes telles que température > 30`

```python
df[df['temperature'] > 30]
```

### `6.2 Récupérer uniquement les lignes telles température > 30 et humidité > 70`

```python
df[(df["temperature"] > 30) & (df["humidite_relative"] > 70)]
```
