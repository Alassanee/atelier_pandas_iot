### `Partie 1 - Series`

#### `1.1 Création d'une Series contenant 4 températures`

```python
import pandas as pd

# Creation de la serie a partir d'une list
temperatures = [25, 34, 23, 40]
serie = pd.Series(temperatures)
serie
```

#### `1.2 Affichage de l'index`

```python
serie.index
```

#### `1.3 Affichage de la 1ère valeur`

```python
serie[0]
```

#### `1.4 Affichage de la dernière valeur`

```python
serie.iloc[-1]
```

#### `1.5 Création, de deux façon différentes, d'une Series de 4 températures avec comme index les heures demesure 12h, 13h, 14h et 15h`

```python
# Methode 1: a partir d'un dictionnaire
dict_temp = {
     "12h": 34,
     "13h": 23,
     "14h": 40,
     "15h": 37
}
serie_1 = pd.Series(dict_temp)
serie_1
```

```python
# Methode 2: A partir d'une liste,
indexes = ["12h", "13h", "14h", "15h"] 
serie_2 = pd.Series(temperatures, index=indexes)
serie_2
```

#### `1.6 Affichage de la 1ère et la dernière valeur`

```python
#
serie_2.iloc[[0, -1]]
```
